Q3 Customer Purchase Analysis

Q7 Pivot Table Creation

Create a pivot table showing:

Rows → Region

Columns → Product

Values → Total Sales

Which region sells the most laptops?

```python
from textwrap import fill
# Pivot Table

Pivot_table = pd.pivot_table(
    df,
    values='Sales',
    index='Region',
    columns='Product',
    aggfunc='sum',
    fill_value=0
)

Pivot_table

# East Region Sells to most laptop
```

Q10 End-to-End Data Wrangling Task

Perform the following workflow:

Filter the dataset where Sales > 3000

Group data by Region and Product

Calculate total sales

Reshape the result into a pivot table

Explain the business insights obtained from this analysis.

```python
# filtered data

filtered_df = df[df['Sales'] > 3000 ]

filtered_df
```

```python
# gruped data

grouped_data = filtered_df.groupby(['Region', 'Product'])['Sales'].sum().reset_index()

grouped_data
```

```python
# Calculate total sales

grouped_data['Sales'].sum()
```

```python
# Filter data
filtered_df = df[df['Sales'] > 3000]

# Group by Region and Product
grouped_data = filtered_df.groupby(['Region', 'Product'])['Sales'].sum().reset_index()

# Create Pivot Table
pivot_table = grouped_data.pivot_table(
    index='Region',
    columns='Product',
    values='Sales',
    fill_value=0
)

print(pivot_table)
```
