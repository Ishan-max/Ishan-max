```python
import pandas as pd

df = pd.read_csv('/content/House Price India.csv')
df
```

Q1. What is the average, median, and standard deviation of house prices?

```python
# Avg, Median and STD



stats = df['Price'].describe()


print("Average (Mean):", stats['mean'])
print("Median:", stats['50%'])
print("Standard Deviation:", stats['std'])

# First we are defining stats value of describe code which shows avg, median, std and other values
# then we are extracting only the required values from the stats variable
```

Q2. Which number of bedrooms is most common?

```python
# To find the most common number of bedrooms in the dataset, we use the mode. The mode is the value that appears most frequently in a column.

most_common_bedroom = df['number of bedrooms'].mode()[0]

print("The most common number of bedrooms is:", most_common_bedroom)
```

Q3. Check if the price data is skewed.

```python
# Skewness helps us understand whether the distribution of house prices is symmetric or tilted toward one side.

price_skewness = df['Price'].skew()

print(price_skewness)
```

Q4. What is the average price for each number of bedrooms?

```python
# To find the average house price for each bedroom category, we use the groupby() function and calculate the mean of the Price column for each unique value in the Bedrooms column.

avg_price_by_bedroom = df.groupby('number of bedrooms')['Price'].mean().round(2)

print(avg_price_by_bedroom)
```

Q5.  What is the relationship between living area and price? (Use : Area without Basement).

```python

# To determine the relationship between Area without Basement and Price, we can calculate the correlation coefficient.

correlation = df['living area'].corr(df['Price'])

print('Correlation between living area and price', correlation)
```

Q6. Identify any anomalies where houses have high prices but low areas.

```python
# Anomalies (outliers) are houses that have unusually high prices despite having a relatively small living area. One simple approach is to identify houses with:
'''
Price greater than the 75th percentile (high price), and
living area less than the 25th percentile (low area). '''

high_price = df['Price'].quantile(0.75)
low_area = df['living area'].quantile(0.25)

anomalies = df[
         (df['Price'] > high_price) &
         (df['living area'] < low_area)
         ][['Price', 'living area']]

print(anomalies)
#
```

Q7. Compare average price based on number of floors and houses with or without waterfront.

```python
# To compare house prices based on both number of floors and waterfront status, we can group the data by these two columns and calculate the average price.

avg_price_df = (df.groupby(['number of floors', 'waterfront present'])['Price'].mean().round(2))

print(avg_price_df)
```

Q8. Identify the minimum and maximum house price. What does this indicate?

```python
# To find the minimum and maximum house prices, use the min() and max() functions on the Price column.

min_price = df['Price'].min()
max_price = df['Price'].max()

price_range_df = pd.DataFrame({'Minimum Price': [min_price], 'Maximum Price': [max_price]})

price_range_df
```

A large gap between the minimum and maximum prices indicates that the housing market contains properties from different price segments, such as affordable homes and luxury properties.

Q9. Which location (zipcode/area) has the highest average price?

```python
# To find the location with the highest average house price, group the data by Zipcode (or Area, depending on your dataset) and calculate the average price for each location.

zipcode_price_df = (
    df.groupby('Postal Code')['Price'].mean().round(2).sort_values(ascending=False).head(1)
)

print(zipcode_price_df)
```
