```python
import pandas as pd
df = pd.read_csv('/content/retail_dataset.csv')
df

df.head()
```

```python
print(df.isnull().sum())
```

```python
# replace null value with mean of the columns

df.fillna({
    'Quantity' : df['Quantity'].mean(),
    'Price' : df['Price'].mean()
}, inplace=True )

print(df)
```

```python
df.dropna(subset=['Product Category', 'Region'], inplace=True)

print(df)
```

```python
print(df.isnull().sum())
```

```python
import numpy as np
import pandas as pd


df['Revenue'] = np.multiply(df['Quantity'], df['Price'])

df
```

```python
total_revenue = np.sum(df['Revenue'])
total_revenue

print('Total Revenue', total_revenue)
```

```python
category_revenue = df.groupby('Product Category')['Revenue'].sum()
print(category_revenue)
```

```python
top_3 = category_revenue.nlargest(3)
print(top_3)



bottom_3 = category_revenue.nsmallest(3)
print(bottom_3)
```

```python
region_revenue = df.groupby('Region', as_index=False)['Revenue'].sum()

Highest_revenue = region_revenue.sort_values('Revenue', ascending=False).head(1)
Lowest_revenue = region_revenue.sort_values('Revenue').head(1)

print(Highest_revenue)
print(Lowest_revenue)
```

```python
import pandas as pd
import numpy as np

# Convert date column to datetime format
df['Date'] = pd.to_datetime(df['Date'], format='%d-%m-%Y')

# Extract month from date column

df['Month'] = df['Date'].dt.month

monthly_revenue = df.groupby('Month', as_index=False)['Revenue'].sum()

print(round(monthly_revenue,2))

# Calculate mean, median and standard deviation of Revenue

mean_revenue = np.mean(df['Revenue'])
median_revenue = np.median(df['Revenue'])
std_revenue = np.std(df['Revenue'])

print("\n Revenue Statisctics:")
print("Mean:", round(mean_revenue,2))
print("Median:",round( median_revenue,2))
print("Standard Deviation:", round(std_revenue,2 ))


```
