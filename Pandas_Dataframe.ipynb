Q9. Descriptive Statistics

Generate summary statistics using:

describe()

From the result identify:

Mean sales value

Maximum quantity sold

Minimum sales value




```python
# Generate Summary

summary = df.describe()

# Mean sales value

Mean_sales = summary.loc['mean', 'Sales']

# Maximum Quantity Sold

Max_Qty = summary.loc['max', 'Quantity']

# Minimum Sales Value

Min_Sales = summary.loc['min', 'Sales']

print("Mean Sales Value:", Mean_sales)
print("Maximum Quantity Sold:", Max_Qty)
print("Minimum Sales Value:", Min_Sales)

```

Q10. Aggregation Analysis

Using the Sales column, calculate:

Total sales

Average sales

Maximum sales

Minimum sales

```python
# Calculate Total Sales, AVG Sales, Maximum sale, Minimum sales,

Total_Sales = df['Sales'].sum()
Avg_Sales = df['Sales'].mean()
Max_Sales = df['Sales'].max()
Min_Sales = df['Sales'].min()

print("Total Sales:", Total_Sales)
print("Average Sales:", Avg_Sales)
print("Maximum Sales:", Max_Sales)
print("Minimum Sales:", Min_Sales)
```
