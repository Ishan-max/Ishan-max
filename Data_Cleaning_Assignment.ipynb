```python
import pandas as pd

df = pd.read_csv('/content/healthcare_data_cleaning_dataset.csv')

df.head()
```

Q1. Missing Data Identification

Scenario:
 The hospital suspects incomplete patient records.

Task:

Identify missing values in each column

Calculate percentage of missing data



```python
missing_values = df.isnull().sum()
missing_values_percentage = (missing_values / len(df)) * 100

print("Missing Values:")
print(missing_values)
print("\nPercentage of Missing Data:")
print(missing_values_percentage)
```

Q2. Handling Missing Age

Scenario:
 Age is critical for medical analysis, but some values are missing.

Task:

Replace missing Age values with an appropriate method

Justify your choice (mean/median)

```python
''' Missing values in the Age column were replaced using the median because age data may contain outliers.
 The median is more robust than the mean and provides a better representation of the central tendency of the data
  without being heavily influenced by extreme values.'''



df['Age'] = df['Age'].fillna(df['Age'].median())

df['Age']
```

Q3. Handling Missing Treatment Cost

Scenario:
 Treatment cost is highly skewed due to expensive treatments.

Task:

Handle missing Treatment_Cost values

Choose the correct imputation method and explain why

```python
''' Handle missing Treatment_Cost values
Choose the correct imputation method and explain why '''

df['Treatment_Cost'] = df['Treatment_Cost'].fillna(df['Treatment_Cost'].median())

df['Treatment_Cost']
```

Q4. Duplicate Patient Records

Scenario:
 Some patient records were entered multiple times.

Task:

Identify duplicate rows

Remove duplicates

Compare dataset size before and after

```python
# Dataset size before
print("Dataset Size Before:", df.shape)

# Identify duplicates
duplicates = df[df.duplicated()]
print("\nDuplicate Rows:")
print(duplicates)

# Remove duplicates
df = df.drop_duplicates()

# Dataset size after
print("\nDataset Size After:", df.shape)
```

Q5. Invalid Age Values (Data Quality Check)

Scenario:
 Some patients have unrealistic age values (e.g., >100 or <0).

Task:

Detect such records

Decide whether to remove or correct them

```python
# find invalid age

# Find records with invalid ages


median_age = df[(df['Age'] >= 0) & (df['Age'] <= 100)]['Age'].median()

df.loc[(df['Age'] < 0) | (df['Age'] > 100), 'Age'] = median_age

print(df)

```

Q6. Outlier Detection (Treatment Cost)

Scenario:
 Extreme treatment costs are affecting analysis.

Task:

Detect outliers using IQR method

Display number of outliers

```python
# Calculate Q1 and Q3

Q1 = df['Treatment_Cost'].quantile(0.25)
Q3 = df['Treatment_Cost'].quantile(0.75)

# Calculate the Interquartile Range (IQR)
IQR = Q3 - Q1

# Define the lower and upper bounds for outliers

lower_band = Q1 - 1.5 * IQR
upper_band = Q3 + 1.5 * IQR

# Identify outliers
outliers = df[
    (df['Treatment_Cost'] < lower_band) | (df['Treatment_Cost'] > upper_band)
]

print("Outlier Records:")
print(outliers)

# Display number of outliers
print("Number of Outliers:", len(outliers))
```

Q7. Outlier Treatment

Scenario:
The business team wants to retain all records.

Task:

Apply capping (Winsorization) on Treatment_Cost

Use 5th and 95th percentile

```python
# calculate 5th and 95th percentiles

lower_limit = df['Treatment_Cost'].quantile(0.05)
upper_limit = df['Treatment_Cost'].quantile(0.95)

# Apply capping on Treatment_Cost
df['Treatment_Cost'] = df['Treatment_Cost'].clip(
    lower=lower_limit,
    upper=upper_limit
)

print(df)

print(lower_limit)
print(upper_limit)

print(df['Treatment_Cost'].describe())
```

Q8. Transformation

Scenario:
 Treatment cost is highly skewed.

Task:

Apply log transformation

Create a new column

Compare before vs after distribution

```python
import numpy as np

# create log-transformed column

df['Log_Treatment_Cost'] = np.log1p(df['Treatment_Cost'])

print(df[['Treatment_Cost', 'Log_Treatment_Cost']].head())



print("Before Transformation")
print(df['Treatment_Cost'].describe())

print("After Transformation")
print(df['Log_Treatment_Cost'].describe())

print("Skewness Before:", df['Treatment_Cost'].skew())
print("Skewness After:", df['Log_Treatment_Cost'].skew())

import matplotlib.pyplot as plt

# Original Distribution

plt.hist(df['Treatment_Cost'], bins = 20 )
plt.title('Treatmeant_Cost_Distribution')
plt.show()

# Log-Transferred distribution

plt.hist(df['Log_Treatment_Cost'], bins = 20)
plt.title('Log_Treatmeant_Cost_Distribution')
plt.show()

```
